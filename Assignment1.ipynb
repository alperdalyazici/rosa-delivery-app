```python
def get_times(promise, zone, time_block):

  # This function returns delivery times of all orders in the chosen zones and time block


  # If the user wants all zones or all time blocks, we need to handle that here.
  
  if zone == 'all':
    zones = ZONES
  else:
    zones = [zone]
  if time_block == 'all':
    blocks = TIME_BLOCKS
  else:
    blocks = [time_block]

  # delivery_times handles one zone and one block at a time, so loop over each pair
  arrays = []
  for z in zones:
    for b in blocks:
      arrays.append(delivery_times(z,b, promise, seed = SEED))

  # Put every order time into one array
  return np.concatenate(arrays)
```

```python
get_times(45, 'North', 'Lunch') # Check that the get_times function works properly
```

```python
len(get_times(45, 'North', 'Lunch')) # Check that the number of orders is correct
```

```python
def late_rate(promise, zone,time_block):
  # This function returns the percentage of orders delivered later than the promise.

  t = get_times(promise, zone, time_block) # Get all the delivery times for the chosen zone and time block
  late_count = np.sum(t > promise)   # how many are late
  total = len(t)                     # how many orders in total
  return round((100 * late_count / total), 2)
```

```python
p = 45  # promised delivery time in minutes

# Print out the late rates for different zones and time blocks to check that the late_rate function works properly.
print(f"Late rate, Far West / Fri/Sat eve (specific zone, specific block), {p}-minute promise: {round(late_rate(p, 'Far West', 'Fri/Sat eve'), 2)}% of orders were late\n")
print(f"Late rate, all zones / Lunch (all zones, specific block), {p}-minute promise: {round(late_rate(p, 'all', 'Lunch'), 2)}% of orders were late\n")
print(f"Late rate, North / all time blocks (specific zone, all blocks), {p}-minute promise: {round(late_rate(p, 'North', 'all'), 2)}% of orders were late\n")
print(f"Late rate, all zones / all time blocks (everything), {p}-minute promise: {round(late_rate(p, 'all', 'all'), 2)}% of orders were late\n")
```

# Part 1 Section B

```python
results = []

# Loop over all zones and time blocks, and calculate the late rate for each combination. Store the results in a list of tuples.
for z in ZONES:
  for b in TIME_BLOCKS:
    results.append((z, b, late_rate(PROMISE, z, b)))


# Sort the results by late rate in descending order, so the highest late rates are at the top.
results.sort(key = lambda x: x[2], reverse= True)


# Print out the results in a nice format, with the rank, zone, time block, and late rate.
print(f'Ranking by late rate, {PROMISE}-minute promise\n')

rank = 1
for i in results:

  print(f"{rank}. {i[0]}, {i[1]}: {i[2]}%\n")
  rank += 1
```

# Part 1 Section C

```python
def avg_time(promise, zone, time_block):
  # This function returns the average delivery time (minutes) of orders in the chosen group.
  t = get_times(promise, zone, time_block)
  return round(np.mean(t), 2)
```

```python
p = 45  # promised delivery time in minutes

# Four cases: specific/specific, all/specific, specific/all, all/all
print(f"Average delivery time, Far West / Fri/Sat eve (specific zone, specific block), {p}-minute promise: {round(avg_time(p, 'Far West', 'Fri/Sat eve'), 2)} minutes\n")
print(f"Average delivery time, all zones / Lunch (all zones, specific block), {p}-minute promise: {round(avg_time(p, 'all', 'Lunch'), 2)} minutes\n")
print(f"Average delivery time, North / all time blocks (specific zone, all blocks), {p}-minute promise: {round(avg_time(p, 'North', 'all'), 2)} minutes\n")
print(f"Average delivery time, all zones / all time blocks (everything), {p}-minute promise: {round(avg_time(p, 'all', 'all'), 2)} minutes\n")
```

# Part 1 Section E

The late rate is more relevant. Rosa's costs come from late orders, which triggers a refund and causes the customer to place fewer orders in the future. She does not pay anything for a slow average, so the late rate tracks her actual costs while the average delivery time does not.
<br><br>

The results from (b) and (d) show how the average can mislead. Far West on Fri/Sat evenings has an average delivery time of 42.92 minutes, which is under the 45-minute promise, so it looks acceptable. Yet 36.67% of its orders are late, the worst rate of any pair. The average hides this because a group can have many on-time orders pulling the average down while a large share still arrive late.
<br><br>

The rankings also disagree in places. Far West Lunch ranks 6th by average time (30.58 minutes) but is tied for last by late rate (2.5%), and it has a higher average than Central Weekday eve (29.03 minutes) despite a lower late rate (2.5% vs 4.24%). Ranking by average time would send Rosa's attention to the wrong places.
<br><br>

Average time still has some use as context, since the slowest pairs are generally the ones with the most late orders. But for deciding where to act and what a promise will cost, the late rate is the better measure.

```python
def get_cost(costs):
  # This function returns the total cost of one late order
  cost = costs['refund'] + (costs['churn_orders'] * costs['margin']) # Calculate the total cost of one late order by looking up the refund, churn_orders, and margin values in the costs dictionary and applying the formula.
  return cost
```

```python
# Quick example to see of the function works properly.
print(f"Cost per late order: ${round(get_cost(COSTS), 2)}")
```

# Part 2 Section B

**Choosing the range of promises**

I will test promises from **20 to 80 minutes in steps of 5**.

- **Lower bound (20 minutes):** A promise below 20 minutes is unrealistic for food delivery. Almost every order would be late, so late costs would be at their highest and net profit would be strongly negative.
- **Upper bound (80 minutes):** I expect 80 minutes to be about the longest wait a customer would accept for delivery. Beyond that, demand would drop too far for the promise to be worth considering.
- **Step size (5 minutes):** A 5-minute step is fine enough to locate the promise with the highest net profit, without testing an excessive number of values.

After running the search, I will check that the best promise is not the first or last value in this range. If it is, the range would need to be extended.

```python
promises = list(range(20,81,5))
```

```python
# Now we need a function that will return the promise with the highest profit margin and the promise number
```

```python
def best_promise(zone, time_block, promises, costs):

  # This function returns the promise with the highest profit margin and the promise number
  late_cost = get_cost(costs)
  best_prom = None
  best_profit = None

  # Loop over all promises and calculate the profit for each one. Keep track of the best promise and best profit.
  for p in promises:
    t = get_times(p, zone, time_block)
    late_count = np.sum(t > p)
    total = len(t)
    profit = total * costs['margin'] - late_count * late_cost

    if best_profit is None or profit > best_profit:
      best_profit = profit
      best_prom = p

  return best_prom, best_profit
#
```

```python
# Quick example to see if the function works properly and returns the best promise and profit for a specific zone and time block.
best_prom, profit = best_promise("Far West", "Fri/Sat eve", promises, COSTS)
print(f"Far West / Fri/Sat eve: best promise = {best_prom} minutes, net profit = ${round(profit, 2)}")
```
